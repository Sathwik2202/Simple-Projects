```python
from matplotlib.image import imread
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [16, 8]  #Setting figure size

A = imread('eye.jpg')
X = np.mean(A, axis=2)
rank = np.linalg.matrix_rank(X)   #Calculating rank

print(np.shape(X))                #Printing dimensions
print('Rank = ', rank)            #Printing rank

img = plt.imshow(X)
img.set_cmap('gray')              #Output image in gray scale
plt.show()
```

```python
U, S, Vt = np.linalg.svd(X, full_matrices = False)
S = np.diag(S)
j = 0
for r in (5, 20, 100):                                 #Image matrix with various sizes(compressed)
    Xapprox = U[:, :r] @ S[0:r, :r] @ Vt[:r,:]
    new_rank = np.linalg.matrix_rank(Xapprox)
    print('rank = ', new_rank)
    plt.figure(j+1)
    j += 1
    img = plt.imshow(Xapprox)
    img.set_cmap('gray')
    plt.title('r =' + str(r))
    plt.show()
    print("")
```
